## 6. Statistical Analysis

A 95% confidence interval is calculated to estimate the range in which the population mean salary may lie.

```python
salary = df_clean["ConvertedCompYearly"].dropna()

mean_salary = salary.mean()
std_salary = salary.std()
n = len(salary)

margin_of_error = 1.96 * (std_salary / np.sqrt(n))

lower = mean_salary - margin_of_error
upper = mean_salary + margin_of_error

print("Mean Salary:", mean_salary)
print("95% Confidence Interval:", lower, "to", upper)
```
